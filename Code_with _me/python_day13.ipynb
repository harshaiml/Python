## String Methods in Python 

```python
# Strings are immutable
a = "Harsh"
print(len(a))
print(a)
print(a.upper())
print(a.lower())
```

```python
# Strings are immutable
a = "Harsh!!!!!!!!!!"
print(len(a))
print(a)
print(a.upper())
print(a.lower())
print(a.rstrip("!"))
```

```python
# Strings are immutable
a = "!!Harsh!!!!!!!!!!"
print(len(a))
print(a)
print(a.upper())
print(a.lower())
print(a.rstrip("!"))
```
