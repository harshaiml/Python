## Dictionary Methods in Python

```python
ep1 = {
    122: 45,
    123: 89,
    567: 69,
    670: 69
}
ep2 = {
    222: 67,
    566: 90
}

ep1.update(ep2)
print(ep1)
```

```python
ep1 = {
    122: 45,
    123: 89,
    567: 69,
    670: 69
}
ep1.clear()
print(ep1)
```

```python
# empty dictionary
empt = {}
print(empt)
```

```python
ep1 = {
    122: 45,
    123: 89,
    567: 69,
    670: 69
}
ep1.pop(122)
print(ep1)
```

```python
ep1 = {
    122: 45,
    123: 89,
    567: 69,
    670: 69
}
ep1.popitem() # Remove last key pair value
print(ep1)
```

```python
ep1 = {
    122: 45,
    123: 89,
    567: 69,
    670: 69
}
del ep1 # Delete ep1
print(ep1)
```

```python
ep1 = {
    122: 45,
    123: 89,
    567: 69,
    670: 69
}
del ep1[122] # Delete 122 key from dictinory
print(ep1)
```
