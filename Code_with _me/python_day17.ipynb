### For Loops in Python

```python
name = 'Abhishek'
for i in name:
    print(i)
```

```python
colors = ["Red", "Green", "Blue", "Yellow"]
for color in colors:
    print(color)
```

```python
colors = ["Red", "Green", "Blue", "Yellow"]
for color in colors:
    print(color)
    for i in color:
        print(i)
```

```python
# Print numbere 1 to 5 using loops
for i in range(5):
    print(i)
```

```python
# print number 1 to 5 using loops
for i in range(5):
    print(i+1)
```

```python
for i in range(1, 11):
    print(i)
```
