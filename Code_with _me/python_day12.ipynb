### String Slicing and Operations on Strings in Python

```python
names = "Krishna,Harsh"
print(names)
print(len(names))
print(names[0:7]) # Krishna
```

```python
names = "Krishna,Harsh"
print(names)
print(len(names))
print(names[:7]) # Krishna
```

```python
names = "Krishna,Harsh"
print(names)
print(len(names))
print(names[8:13]) # Harsh
```

```python
names = "Krishna,Harsh"
print(names)
print(len(names))
print(names[:]) # Harsh
```

### Negative Slicing

```python
fruit = "Mango"
print(fruit[-1:-6:-1]) # Start, Stop, Step
```

```python
name = "harsh"
print(name[-4:-2])
```
