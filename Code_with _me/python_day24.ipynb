## Tuples in Python

```python
tup = (1, 5, 6)
print(type(tup), tup)
```

```python
tup = (1,)
print(type(tup), tup)
```

```python
tup = (1, 2, 76, 342, 32, "green", True)
print(type(tup), tup)
```

```python
tup = (1, 2, 76, 342, 32, "green", True)
print(type(tup), tup)
print(tup[0])
print(tup[1])
print(tup[2])
print(tup[3])
print(tup[-1])
```

```python
tup = (1, 2, 76, 342, 32, "green", True)
print(type(tup), tup)
print("Lenght of tu is:",len(tup))
# print(tup[0])
# print(tup[1])
# print(tup[2])
# print(tup[3])
# print(tup[-1])
```

```python
tup = (1, 2, 76, 342, 32, "green", True)

if 342 in tup:
    print("Yes 342 is present in this tuple")
```

```python
tup = (1, 2, 76, 342, 32, "green", True)
tup2 = tup [1:4]
print(tup2)
```
