```python
name = "Harsh" # String
print(name[0]) # Indexing
print(name[1]) # Indexing
print(name[2]) # Indexing
print(name[3]) # Indexing
print(name[4]) # Indexing
# print(name[5]) # Throw an Error
```

### Lets use a for loop

```python
name = "Harsh" # String
for character in name:
    print(character)
```
