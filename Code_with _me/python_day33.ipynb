## Dictionaries in Python

```python
dic = {
    "Harsh": "Human being",
    "Spoon": "Object"
}
print(dic["Harsh"])
```

```python
dic = {
    344: "Harsh",
    561: "Subham",
    678: "Diviya",
    567: "Tiya"
}
print(dic[567])
```

```python
info = {
    'Name': "Tiya",
    'Age': "22",
    'Add': "Jawan",
    'Eligible': True
    }
print(info)
```

```python
info = {
    'Name': "Tiya",
    'Age': "22",
    'Add': "Jawan",
    'Eligible': True
    }
print(info)
```

```python
info = {
    'Name': "Tiya",
    'Age': "22",
    'Add': "Jawan",
    'Eligible': True
    }
print(info)
print(info['Name'])
print(info['Age'])
print(info['Add'])
print(info['Eligible'])
```

```python
info = {
    'Name': "Tiya",
    'Age': "22",
    'Add': "Jawan",
    'Eligible': True
    }
print(info)
print(info.get('Name'))
print(info.get('Age'))
print(info.get('Add'))
print(info.get('Eligible'))
```

```python
info = {
    'Name': "Tiya",
    'Age': "22",
    'Add': "Jawan",
    'Eligible': True
    }
print(info)
print(info.keys())
```

```python
info = {
    'Name': "Tiya",
    'Age': "22",
    'Add': "Jawan",
    'Eligible': True
    }
print(info)
print(info.keys())

for key in info.keys():
    print(info[key])
```

```python
info = {
    'Name': "Tiya",
    'Age': "22",
    'Add': "Jawan",
    'Eligible': True
    }
print(info)
print(info.values())

for key in info.keys():
    print(info[key])
```

```python
info = {
    'Name': "Tiya",
    'Age': "22",
    'Address': "Jawan",
    'Eligible': True
    }
print(info)
print(info.values())

for key in info.keys():
    print(f"The value corresponding is the key {key} is {info[key]}")
```

```python
info = {
    'Name': "Tiya",
    'Age': "22",
    'Address': "Jawan",
    'Eligible': True
    }
print(info.items())
```

```python
info = {
    'Name': "Tiya",
    'Age': "22",
    'Address': "Jawan",
    'Eligible': True
    }
print(info.items())
for key, value in info.items():
    print(f"The value corresponding to the key {key} is {value}")
```
