### List Methods in Python

```python
l = [1 , 2, 4, 6]
print(l)
```

```python
l = [1 , 2, 4, 6]
l.append(7)
print(l)
```

```python
l = [11, 45, 1 , 2, 4, 6]
l.sort()
print(l)
```

```python
l = [11, 45, 1 , 2, 4, 6]
l.sort(reverse=True)
print(l)
```

```python
l = [11, 45, 1 , 2, 4, 6]
l.reverse()
print(l)
```

```python
l = [11, 45, 1 , 2, 4, 6]
print(l.index(1))
```

```python
l = [11, 45, 1 , 2, 4, 6, 1]
print(l.index(1))
```

```python
l = [11, 45, 1 , 2, 4, 6, 1, 1]
print(l.count(1))
```

```python
l = [11, 45, 1 , 2, 4, 6, 1, 1]
m = l
m[0] = 0
print(l)
```

```python
l = [11, 45, 1 , 2, 4, 6, 1, 1]
m = l.copy()
m[0] = 0
print(l)
```

```python
l = [11, 45, 1 , 2, 4, 6, 1, 1]
l.insert(1, 899)
print(l)
```

```python
l = [11, 45, 1 , 2, 4, 6, 1, 1]
m = [900, 1000, 1100]
l.extend(m)
print(l)
```

```python
l = [11, 45, 1 , 2, 4, 6, 1, 1]
m = [900, 1000, 1100]
k = l + m
print(k)
```
