### Introduction to list in Python

```python
l = [3, 5, 6]
print(l)
print(type(l))
```

```python
marks = [3, 5, 6]
print(marks)
print(type(marks))
print(marks[0])
print(marks[1])
print(marks[2])
```

```python
# Multiple data type in list
marks = [3, 5, 6, "Harsh", True]
print(marks)
print(type(marks))
print(marks[0])
print(marks[1])
print(marks[2])
print(marks[3])
print(marks[4])
# print(marks[4]) # Error
```

```python
colors = ["Red", "Green", "Blue", "Yellow", "Aqua"]
print(colors[2])
```

```python
# Multiple data type in list
marks = [3, 5, 6, "Harsh", True]
print(marks[-3]) # Negative index
print(marks[len(marks)-3]) # Positive index
print(marks[5-3]) # Positive index
print(marks[2]) # Positive Index
```

```python
marks = [3, 5, 6, "Harsh", True]
if 7 in marks:
    print("Yes")
else:
    print("No")
```

```python
marks = [3, 5, 6, "Harsh", True]
if "Harsh" in marks:
    print("Yes")
else:
    print("No")
```

```python
marks = [3, 5, 6, "Harsh", True]
if "Har" in "Harsh":
    print("Yes")
else:
    print("No")
```

```python
marks = [3, 5, 6, "Harsh", True]
print(marks)
print(marks[:])
```

```python
marks = [3, 5, 6, "Harsh", True]
print(marks)
print(marks[1:])
```

```python
marks = [3, 5, 6, "Harsh", True]
print(marks)
print(marks[1:-1])
print(marks[1:4])
```

```python
marks = [3, 5, 6, "Harsh", True]
print(marks)
print(marks[1:-1])
print(marks[1:4:2])
```

```python
marks = [3, 5, 6, "Harsh", True, 6, 7, 2, 32, 345, 23]
print(marks)
print(marks[1:8])
print(marks[1:8:2])
```

```python
marks = [3, 5, 6, "Harsh", True, 6, 7, 2, 32, 345, 23]
print(marks)
print(marks[1:9])
print(marks[1:9:2])
```

### List Comprehension

```python
lst = [i*i for i in range(10)]
print(lst)
```

```python
lst = [i*i for i in range(10) if i%2==0]
print(lst)
```
