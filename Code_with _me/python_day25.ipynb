## Operations on Tuple in Python

Tuple  are immutable, hence if you want to add, remove or change tuple items, them frst you must convert the tuple to a list. Then perform operation on the list and convert it back to tuple.

```python
countries = ("spain", "Italy", "India", "Egland", "Germany")
temp = list(countries)  # Convert into list
temp.append("Russia")   # Add new country
temp.pop(3)             # Remove a country
temp[2] = "Finland"     # Change item
countries = tuple(temp)
print(countries)
```

```python
countries = ("Pakistan", "Afgahanistan", "Bangladesh", "Shrilanka")
countries2 = ("Vietnam", "India", "China")
soutEaastAisa = countries + countries2
print(soutEaastAisa)
```

```python
tuple1 = (0, 1, 2, 3, 2, 3, 1, 3, 2, 3)
res = tuple1.count(3)
print("Count of 3 in tuple is:",res)
```

```python
tuple1 = (0, 1, 2, 5, 2, 3, 1, 3, 2, 3)
res = tuple1.index(3)
print("index is:",res)
```
