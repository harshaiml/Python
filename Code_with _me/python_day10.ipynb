```python
a = input("Enter your 1st number")
b = input("Enter your 2nd number")
sum = (int(a) + int(b))
sub = (int(a) - int(b))
multi = (int(a) * int(b))
dev = (int(a) / int(b))
print(f"You enter the 1st no is {a}")
print(f"You enter the 2nd no is {b}")
print("Sum of 2 numebers", sum)
print("Subtraction of 2 numebers", sub)
print("Multiplication of 2 numebers", multi)
print("Devision of 2 numebers", dev)
```
